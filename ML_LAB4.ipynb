```python
# Experiment 4: Candidate-Elimination Algorithm

import csv

# Read dataset
filename = input("Enter CSV file name: ")

with open(filename, 'r') as file:
    data = list(csv.reader(file))

header = data[0]
examples = data[1:]

attributes = header[:-1]
n = len(attributes)

# Most specific hypothesis
S = ['Ø'] * n

# Most general hypothesis
G = [['?'] * n]


# Check whether hypothesis covers example
def covers(h, example):
    for i in range(n):
        if h[i] != '?' and h[i] != example[i]:
            return False
    return True


# Generalize S
def generalize_S(S, example):
    new_S = S.copy()

    for i in range(n):
        if new_S[i] == 'Ø':
            new_S[i] = example[i]
        elif new_S[i] != example[i]:
            new_S[i] = '?'

    return new_S


# Process examples
for row in examples:

    x = row[:-1]
    label = row[-1].lower()

    if label == "yes":

        # Remove hypotheses from G that don't cover positive example
        G = [g for g in G if covers(g, x)]

        # Generalize S
        S = generalize_S(S, x)

    else:

        # Remove hypotheses from G that cover negative example
        new_G = []

        for g in G:
            if covers(g, x):

                # Generate minimal specializations
                for i in range(n):
                    if g[i] == '?' and S[i] != x[i]:
                        h = g.copy()
                        h[i] = S[i]
                        new_G.append(h)
            else:
                new_G.append(g)

        G = new_G


print("\nAttributes:", attributes)

print("\nMost Specific Hypothesis (S):")
print(S)

print("\nMost General Hypotheses (G):")
for g in G:
    print(g)

print("\nFinal Version Space:")
print("S =", S)
print("G =", G)
```

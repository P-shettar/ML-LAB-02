```python
# Experiment 3: Find-S Algorithm

import csv

# Read dataset
filename = input("Enter CSV file name: ")

with open(filename, 'r') as file:
    data = list(csv.reader(file))

# Separate header and examples
header = data[0]
examples = data[1:]

# Last column is the target/class
attributes = header[:-1]

# Find first positive example
hypothesis = None

for row in examples:
    if row[-1].lower() == "yes":
        hypothesis = row[:-1].copy()
        break

# Apply Find-S
for row in examples:
    if row[-1].lower() == "yes":
        for i in range(len(hypothesis)):
            if hypothesis[i] != row[i]:
                hypothesis[i] = "?"

# Display final hypothesis
print("\nAttributes:", attributes)
print("Final Hypothesis:", hypothesis)
```

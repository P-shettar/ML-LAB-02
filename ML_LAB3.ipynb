```python
import csv

def find_s(data):

    # Number of attributes
    num_attributes = len(data[0]) - 1

    # Most specific hypothesis
    hypothesis = ['0'] * num_attributes

    # Process ALL rows
    for row in data:

        # Last column is target
        target = row[-1].strip().lower()

        # Find-S considers only positive examples
        if target == 'yes':

            print("\nProcessing Positive Example:", row)

            for i in range(num_attributes):

                # First positive example
                if hypothesis[i] == '0':
                    hypothesis[i] = row[i]

                # Different values -> generalize to ?
                elif hypothesis[i] != row[i]:
                    hypothesis[i] = '?'

            print("Current Hypothesis:", hypothesis)

    return hypothesis


print("======================================")
print("          FIND-S ALGORITHM")
print("======================================")

filename = input("Enter CSV file name: ")

try:

    with open(filename, 'r') as file:

        reader = csv.reader(file)
        data = list(reader)

    print("\nDataset:")

    for row in data:
        print(row)

    final_hypothesis = find_s(data)

    print("\n======================================")
    print("          FINAL HYPOTHESIS")
    print("======================================")

    print("<" + ", ".join(final_hypothesis) + ">")

except FileNotFoundError:

    print("Error: File not found.")

except Exception as e:

    print("Error:", e)
```

```python
import csv


def find_s(data):

    # Mushroom dataset:
    # Column 0 = Class
    # e = Edible (Positive)
    # p = Poisonous (Negative)

    num_attributes = len(data[0]) - 1

    # Start with most specific hypothesis
    hypothesis = ['0'] * num_attributes

    positive_count = 0

    # Process every row
    for row in data:

        # First column is the target
        target = row[0].strip().lower()

        # Find-S considers only positive examples
        if target == 'e':

            positive_count += 1

            # Attributes start from column 1
            attributes = row[1:]

            print("\nPositive Example", positive_count)
            print("Class:", row[0])

            # First positive example
            if hypothesis == ['0'] * num_attributes:
                hypothesis = attributes.copy()

            else:
                # Compare attributes
                for i in range(num_attributes):

                    if hypothesis[i] != attributes[i]:
                        hypothesis[i] = '?'

            print("Current Hypothesis:")
            print("<" + ", ".join(hypothesis) + ">")

    return hypothesis


# ==========================================
#              MAIN PROGRAM
# ==========================================

print("=" * 55)
print("                  FIND-S ALGORITHM")
print("=" * 55)

filename = input("Enter CSV file name: ")

try:

    # Read CSV
    with open(filename, 'r', newline='') as file:

        reader = csv.reader(file)
        data = list(reader)

    # Remove header if present
    if data[0][0].lower() == 'class':
        header = data[0]
        data = data[1:]
    else:
        header = None

    print("\nDataset loaded successfully!")
    print("Total examples:", len(data))
    print("Total attributes:", len(data[0]) - 1)

    # Run Find-S
    final_hypothesis = find_s(data)

    # ==========================================
    #          FINAL HYPOTHESIS
    # ==========================================

    print("\n" + "=" * 55)
    print("                 FINAL HYPOTHESIS")
    print("=" * 55)

    print("<" + ", ".join(final_hypothesis) + ">")

    print("\nMeaning:")
    print("e = Edible (Positive Example)")
    print("p = Poisonous (Negative Example)")
    print("? = Any value")

except FileNotFoundError:

    print("\nERROR: File not found.")

except Exception as e:

    print("\nERROR:", e)
```
